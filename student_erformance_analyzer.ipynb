# 🎓 Student Performance Analyzer

A beginner-friendly Python project that analyzes a student's academic performance and generates a simple performance report.

### 🔍 What This Project Does

* Calculates the student's average marks
* Assigns a grade based on the average
* Counts failed subjects
* Determines the student's overall pass/fail status
* Identifies the highest-scoring subject
* Checks whether the student needs an academic warning
* Generates a final formatted report

### 🛠️ Python Concepts Used

`Functions` · `Lists` · `Loops` · `Conditional Statements` · `Variables` · `Return Values` · `String Formatting`

**Built with Python 🐍**


## 1. Student Information

```python
name = "Hameed"
roll_no = "24AI-30"
```

## 2. Student Data

```python
marks = [92, 45, 37, 91]
subjects = ["Python", "AI", "Networking", "Mathematics"]
```

## 3. Calculate Average

```python
def calculate_average(marks):
  average = sum(marks)/len(marks)
  return average
average = calculate_average(marks)
```

## 4. Calculate Grade

```python
def calculate_grade(average):
  if average >= 90:
    grade = "A"
  elif average >= 80:
    grade = "B"
  elif average >= 70:
    grade = "C"
  elif average >= 60:
    grade = "D"
  else:
    grade = "F"
  return grade
grade = calculate_grade(average)

```

## 5. Check Failed Subjects

```python
#Checking if Any sub Failed!!
def check_subject_failures(marks):
  failed_count = 0
  for i in range(len(marks)):
      if marks[i] < 50:
        failed_count += 1

  return failed_count
failed_count = check_subject_failures(marks)

```

## 6. Check Student Status

```python
def check_status(average, failed_count):
  if average >= 60 and failed_count == 0:
    status = "PASS"
  else:
    status = "FAIL"
  return status
status = check_status(average, failed_count)
```

## 7. Check Academic Warning

```python
def check_academic_warning(average, failed_count):
  if average <70 or failed_count >= 2:
    return "Yes"
  else:
     return "No"
warning = check_academic_warning(average, failed_count)
```

## 8. Find Highest Subject

```python
def find_highest_subject(subjects,marks):
  highest_marks = max(marks)
  highest_index = marks.index(highest_marks)
  highest_subject = subjects[highest_index]
  return highest_subject,  highest_marks
highest_subject, highest_marks = find_highest_subject(subjects,marks)
```

## 9. Generate Final Report

```python
def generate_report(name, roll_no, subjects, marks, average, grade, status, highest_subject, highest_marks, failed_count, warning):
  report = f"Name: {name}\nRoll No: {roll_no}\n\n"
  for i in range(len(subjects)):
    report += f"{subjects[i]}: {marks[i]}\n"
  report += f"Average: {average}\n"
  report += f"Grade: {grade}\n"
  report += f"Status: {status}\n"
  report += f"Highest Subject: {highest_subject}\n"
  report += f"Highest Marks: {highest_marks}\n"
  report += f"Failed Subjects: {failed_count}\n"
  report += f"Academic Warning: {warning}"
  return report
report = (generate_report(name, roll_no, subjects,marks, average, grade, status, highest_subject, highest_marks, failed_count, warning))
print(report)
```
